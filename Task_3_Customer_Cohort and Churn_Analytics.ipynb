## Step 3: Customer Churn Analysis Using Python

```python
## 1. Calculate total customers

total_customers = df["customerID"].nunique()

print("Total Customers:", total_customers)
```

```python
## 2. Calculate churned customers


churned_customers = (df["Churn"] == "Yes").sum()
print("Churned Customers:", churned_customers)
```

```python
## 3. Calculate customers who stayed

retained_customers = (df["Churn"] == "No").sum()

print("Retained Customers:", retained_customers)
```

```python
## 4. Calculate overall churn rate

churn_rate = (
    churned_customers / total_customers
) * 100

print("Churn Rate:", round(churn_rate, 2), "%")
```

```python
## 5. Calculate retention rate

retention_rate = (
    retained_customers / total_customers
) * 100

print("Retention Rate:", round(retention_rate, 2), "%")
```

```python
## 6. Create a summary table

churn_summary = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Churned Customers",
        "Retained Customers",
        "Churn Rate (%)",
        "Retention Rate (%)"
    ],

    "Value": [
        total_customers,
        churned_customers,
        retained_customers,
        round(churn_rate, 2),
        round(retention_rate, 2)
    ]
})

display(churn_summary)
```

## Step 4: Tenure Cohort Analysis (Customer Churn)

```python
## 1. Create tenure groups

bins = [-1, 12, 24, 36, 48, 60, 72]

labels = [
    "0-12 months",
    "13-24 months",
    "25-36 months",
    "37-48 months",
    "49-60 months",
    "61-72 months"
]

df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=bins,
    labels=labels
)

df[["customerID", "tenure", "TenureGroup"]].head(10)
```

```python
## 2. Calculate churn by tenure group

tenure_summary = df.groupby(
    "TenureGroup",
    observed=False
).agg(
    Total_Customers=("customerID", "nunique"),
    Churned_Customers=("Churn", lambda x: (x == "Yes").sum())
)

tenure_summary["Churn Rate (%)"] = (
    tenure_summary["Churned_Customers"] /
    tenure_summary["Total_Customers"] * 100
).round(2)

tenure_summary["Retention Rate (%)"] = (
    100 - tenure_summary["Churn Rate (%)"]
).round(2)

display(tenure_summary)
```

```python
## 3. Create a churn rate chart

plt.figure(figsize=(10, 5))

sns.barplot(
    data=tenure_summary.reset_index(),
    x="TenureGroup",
    y="Churn Rate (%)"
)

plt.title("Customer Churn Rate by Tenure Group")
plt.xlabel("Customer Tenure")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=30)
plt.show()
```

## Step 5: Analyze Churn by Customer Services

```python
## 1. Churn by Contract Type

contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

contract_churn = contract_churn.round(2)

display(contract_churn)


## Create a chart:

contract_churn.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Churn")
plt.show()
```

```python
## 2. Churn by Payment Method

payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

payment_churn = payment_churn.round(2)

display(payment_churn)


## Create a chart:

payment_churn.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Customer Churn by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=30, ha="right")
plt.legend(title="Churn")
plt.show()
```

```python
## 3. Churn by Internet Service

internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

internet_churn = internet_churn.round(2)

display(internet_churn)


## Create a chart:

internet_churn.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Customer Churn by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Churn")
plt.show()
```

## Step 6: Customer Lifetime Value (CLV) Analysis

```python
## 1. Calculate estimated customer value

df["Estimated_CLV"] = df["MonthlyCharges"] * df["tenure"]

df[["customerID", "tenure", "MonthlyCharges",
    "Estimated_CLV", "Churn"]].head(10)
```

```python
## 2. Find high-value customers

clv_median = df["Estimated_CLV"].median()

print("Median Customer Value:", round(clv_median, 2))


## Now create customer groups:

df["CustomerValueGroup"] = np.where(
    df["Estimated_CLV"] >= clv_median,
    "High Value",
    "Low Value"
)

df["CustomerValueGroup"].value_counts()
```

```python
##  3. Analyze churn among high-value customers

high_value_churn = pd.crosstab(
    df["CustomerValueGroup"],
    df["Churn"],
    normalize="index"
) * 100

display(high_value_churn.round(2))
```

```python
## 4. Identify high-value customers who left

high_value_churned = df[
    (df["CustomerValueGroup"] == "High Value") &
    (df["Churn"] == "Yes")
]

high_value_churned[
    ["customerID", "tenure", "MonthlyCharges",
     "Estimated_CLV", "Contract", "PaymentMethod"]
].head(10)
```

```python
## 5. Create a chart

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="CustomerValueGroup",
    hue="Churn"
)

plt.title("Customer Churn by Customer Value Group")
plt.xlabel("Customer Value Group")
plt.ylabel("Number of Customers")
plt.show()
```
